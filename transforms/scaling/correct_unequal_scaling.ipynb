# Correct Unequal Scaling
This can be done by using `save` and `restore`

```python
import cairo
import math
from IPython.core.display import Image

surface = cairo.ImageSurface(cairo.FORMAT_RGB24, 800, 800)
ctx = cairo.Context(surface)
ctx.set_source_rgb(0.8, 0.8, 0.8)
ctx.paint()

cx, cy = 300, 350 # center of enlargement

# Apply transformations and draw shape first
ctx.save()
ctx.translate(cx, cy)
ctx.scale(2, 1)
ctx.translate(-cx, -cy)

ctx.arc(cx, cy, 100, 0, math.pi*2)
ctx.restore()

# Fill and stroke after
ctx.set_source_rgb(0.8, 0, 0)
ctx.fill_preserve()
ctx.set_source_rgb(0, 0, 0)
ctx.set_line_width(10)
ctx.stroke()

surface.write_to_png('ellipse.png')
Image('ellipse.png')
```
