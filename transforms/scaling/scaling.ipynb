# Scaling
It enlarges or shrinks the user space by the specified values

```python
import cairo
import math
from IPython.core.display import Image

surface = cairo.ImageSurface(cairo.FORMAT_RGB24, 600, 600)
ctx = cairo.Context(surface)
ctx.set_source_rgb(0.8, 0.8, 0.8)
ctx.paint()

ctx.scale(1, 2) # Unequal scaling
ctx.arc(300, 150, 100, 0, math.radians(360))
ctx.set_source_rgb(0, 1, 1)
ctx.fill_preserve()
ctx.set_source_rgb(1, 0, 1)
ctx.set_line_width(5)
ctx.stroke()

surface.write_to_png("scaling.png")
Image("scaling.png")
```
