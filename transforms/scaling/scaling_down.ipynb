# Scaling Down
This is equivalent to shrinking.

```python
import cairo
from IPython.core.display import Image

surface = cairo.ImageSurface(cairo.FORMAT_RGB24, 600, 600)
ctx = cairo.Context(surface)
ctx.set_source_rgb(0.8, 0.8, 0.8)
ctx.paint()

width, height = 100, 100

ctx.set_source_rgb(1, 0, 0)
ctx.rectangle(150, 200, width, height)
ctx.fill()

ctx.scale(2/3, 2/3)
ctx.set_source_rgb(0, 0, 1)
ctx.rectangle(300, 150, width, height)
ctx.fill()

surface.write_to_png("down_scaling.png")
Image("down_scaling.png")
```
