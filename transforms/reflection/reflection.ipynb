# Reflection
Flipping the user space across a given line

## Reflection along y-axis
This involves scaling by `(-1, 1)` negating all values of x and translating by `(-width, 0)` where `width` represents the **surface width**.

```python
""" Reflection along y-axis """
import cairo
from IPython.core.display import Image

surface = cairo.ImageSurface(cairo.FORMAT_RGB24, 600, 600)
ctx = cairo.Context(surface)
ctx.set_source_rgb(0.8, 0.8, 0.8)
ctx.paint()

triangle_coordinates = [(100, 200), (100, 400), (200, 300)]

# Triangle before reflection
ctx.move_to(*triangle_coordinates[0])
ctx.line_to(*triangle_coordinates[1])
ctx.line_to(*triangle_coordinates[2])
ctx.close_path()
ctx.set_source_rgb(1, 0, 0)
ctx.stroke()

# Flip user space along y-axis
ctx.scale(-1, 1)
ctx.translate(-600, 0)

# Triangle after reflection
ctx.move_to(*triangle_coordinates[0])
ctx.line_to(*triangle_coordinates[1])
ctx.line_to(*triangle_coordinates[2])
ctx.close_path()
ctx.set_source_rgb(1, 0, 0)
ctx.fill()

ctx.move_to(300, 0)
ctx.line_to(300, 600)
ctx.close_path()
ctx.set_source_rgb(0, 0, 0)
ctx.set_line_width(5)
ctx.set_dash([5, 5])
ctx.stroke()

surface.write_to_png("reflection_along_y_axis.png")
Image("reflection_along_y_axis.png")
```

## Reflection along the x-axis
This involves scaling by `(1, -1)` negating all values of y and translating by `(0, -height)` where `height` represents the **surface height.**


```python
""" Reflection along x-axis """
import cairo
from IPython.core.display import Image

surface = cairo.ImageSurface(cairo.FORMAT_RGB24, 600, 600)
ctx = cairo.Context(surface)
ctx.set_source_rgb(0.8, 0.8, 0.8)
ctx.paint()

triangle_coordinates = [(200, 100), (300, 200), (400, 100)]

# Triangle before reflection
ctx.move_to(*triangle_coordinates[0])
ctx.line_to(*triangle_coordinates[1])
ctx.line_to(*triangle_coordinates[2])
ctx.close_path()
ctx.set_source_rgb(1, 0, 0)
ctx.stroke()

# Flip user space along y-axis
ctx.scale(1, -1)
ctx.translate(0, -600)

# Triangle after reflection
ctx.move_to(*triangle_coordinates[0])
ctx.line_to(*triangle_coordinates[1])
ctx.line_to(*triangle_coordinates[2])
ctx.close_path()
ctx.set_source_rgb(1, 0, 0)
ctx.fill()

ctx.move_to(0, 300)
ctx.line_to(600, 300)
ctx.close_path()
ctx.set_source_rgb(0, 0, 0)
ctx.set_line_width(5)
ctx.set_dash([5, 5])
ctx.stroke()

surface.write_to_png("reflection_along_x_axis.png")
Image("reflection_along_x_axis.png")
```
